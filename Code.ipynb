```python
# ============================================================
# ConvNeXt-Base + EffNetV2_M + Swin-B
# EMA + MixUp + CutMix + Label Smoothing
# K-Fold CV + Feature-fusion ensemble + TTA
# ============================================================
import os
import sys
import glob
import copy
import random
import math
from collections import Counter
from pprint import pprint

import cv2
import numpy as np
import pandas as pd
from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

from tqdm import tqdm
import albumentations as A
from albumentations.pytorch import ToTensorV2
from torchvision import models

# Try to import SAM from torch-optimizer (soft dependency)
try:
    import torch_optimizer as torch_optim_extra
    SAM_AVAILABLE = hasattr(torch_optim_extra, "SAM")
except Exception:
    SAM_AVAILABLE = False

# ============================================================
# Config - tweak these
# ============================================================
BASE_DIR = "/kaggle/input/covid19-radiography-database/COVID-19_Radiography_Dataset"
CLASSES = ["COVID", "Lung_Opacity", "Normal", "Viral Pneumonia"]
NUM_CLASSES = len(CLASSES)
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

SAVE_DIR = "models_99_attempt"
os.makedirs(SAVE_DIR, exist_ok=True)

SEED = 42
BATCH_SIZE = 16       # reduce if OOM
NUM_WORKERS = 4
EPOCHS = 32
N_FOLDS = 1
MIXUP_ALPHA = 0.8
CUTMIX_ALPHA = 1.0
P_MIXUP = 0.6
P_CUTMIX = 0.4
EARLY_STOPPING_PATIENCE = 6
EMA_DECAY = 0.999
LOW_CONTRAST_STD_THRESHOLD = 10.0   # remove images with std < this
TTA_ROUNDS = 4

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if DEVICE == "cuda":
    torch.cuda.manual_seed_all(SEED)

print("Device:", DEVICE, "SAM available:", SAM_AVAILABLE)

# ============================================================
# Build dataframe with only images with masks
# ============================================================
data = []
for label in CLASSES:
    img_paths = glob.glob(f"{BASE_DIR}/{label}/images/*.png")
    for img_path in img_paths:
        mask_path = img_path.replace("images", "masks")
        if os.path.exists(mask_path):
            data.append((img_path, mask_path, label))
df = pd.DataFrame(data, columns=["image_path", "mask_path", "label"])
print("Total samples with masks:", len(df))
pprint(df['label'].value_counts().to_dict())

# ============================================================
# Utilities: mask application + low-contrast filtering
# ============================================================
def apply_mask_and_clahe(img_path, mask_path):
    img = cv2.imread(img_path)
    if img is None:
        raise FileNotFoundError(img_path)
    mask = cv2.imread(mask_path)
    if mask is None:
        # fallback to original
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        return img_rgb

    mask = cv2.resize(mask, (img.shape[1], img.shape[0]))
    mask_gray = cv2.cvtColor(mask, cv2.COLOR_BGR2GRAY)
    _, mask_bin = cv2.threshold(mask_gray, 1, 255, cv2.THRESH_BINARY)
    kernel = np.ones((5,5), np.uint8)
    mask_bin = cv2.morphologyEx(mask_bin, cv2.MORPH_CLOSE, kernel)
    masked = cv2.bitwise_and(img, img, mask=mask_bin)
    blended = cv2.addWeighted(img, 0.2, masked, 0.8, 0)
    lab = cv2.cvtColor(blended, cv2.COLOR_BGR2LAB)
    l, a, b = cv2.split(lab)
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
    l = clahe.apply(l)
    lab = cv2.merge((l,a,b))
    out = cv2.cvtColor(lab, cv2.COLOR_LAB2RGB)
    return out

def is_low_contrast(img_np, std_thresh=LOW_CONTRAST_STD_THRESHOLD):
    # img_np expected RGB uint8
    return img_np.std() < std_thresh

# Filter low contrast images (optional but recommended)
clean_rows = []
removed = 0
for _, row in df.iterrows():
    try:
        img_np = apply_mask_and_clahe(row.image_path, row.mask_path)
        if is_low_contrast(img_np):
            removed += 1
            continue
        clean_rows.append(row)
    except Exception:
        # if reading fails, skip
        removed += 1
        continue

df_clean = pd.DataFrame(clean_rows).reset_index(drop=True)
print(f"Removed {removed} low-quality/failed samples. Remaining: {len(df_clean)}")
pprint(df_clean['label'].value_counts().to_dict())

# ============================================================
# Albumentations transforms (train / val / tta)
# ============================================================
train_transforms = A.Compose([
    A.Resize(224,224),
    A.OneOf([
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.01),
    ], p=0.5),
    A.Affine(scale=(0.9,1.1), translate_percent=(0.05,0.05), rotate=(-20,20), p=0.7),
    A.RandomBrightnessContrast(p=0.5),
    A.CLAHE(p=0.5),
    A.OneOf([A.GaussNoise(var_limit=(10,50)), A.GaussianBlur(blur_limit=3)], p=0.4),
    A.CoarseDropout(max_holes=1, hole_height=32, hole_width=32, p=0.2),
    A.Normalize(mean=(0.485,0.456,0.406), std=(0.229,0.224,0.225)),
    ToTensorV2()
])

val_transforms = A.Compose([
    A.Resize(224,224),
    A.Normalize(mean=(0.485,0.456,0.406), std=(0.229,0.224,0.225)),
    ToTensorV2()
])

tta_transforms = [
    A.Compose([A.Resize(224,224), A.Normalize((0.485,0.456,0.406),(0.229,0.224,0.225)), ToTensorV2()]),
    A.Compose([A.HorizontalFlip(p=1.0), A.Resize(224,224), A.Normalize((0.485,0.456,0.406),(0.229,0.224,0.225)), ToTensorV2()]),
    A.Compose([A.Rotate(limit=15, p=1.0), A.Resize(224,224), A.Normalize((0.485,0.456,0.406),(0.229,0.224,0.225)), ToTensorV2()]),
    A.Compose([A.Rotate(limit=-15, p=1.0), A.Resize(224,224), A.Normalize((0.485,0.456,0.406),(0.229,0.224,0.225)), ToTensorV2()])
]

# ============================================================
# Dataset class
# ============================================================
class LungXRayDataset(Dataset):
    def __init__(self, df, transform=None):
        self.df = df.reset_index(drop=True)
        self.transform = transform
        self.label2idx = {label: idx for idx, label in enumerate(CLASSES)}

    def __len__(self): return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img = apply_mask_and_clahe(row.image_path, row.mask_path)  # RGB numpy
        label = self.label2idx[row.label]
        if self.transform:
            augmented = self.transform(image=img)
            img_t = augmented["image"]
        else:
            img_t = torch.from_numpy(img.transpose(2,0,1)).float() / 255.0
        return img_t, label

# ============================================================
# MixUp & CutMix functions
# ============================================================
def rand_bbox(size, lam):
    W = size[2]
    H = size[3]
    cut_rat = np.sqrt(1. - lam)
    cut_w = int(W * cut_rat)
    cut_h = int(H * cut_rat)
    # uniform
    cx = np.random.randint(W)
    cy = np.random.randint(H)
    bbx1 = np.clip(cx - cut_w // 2, 0, W)
    bby1 = np.clip(cy - cut_h // 2, 0, H)
    bbx2 = np.clip(cx + cut_w // 2, 0, W)
    bby2 = np.clip(cy + cut_h // 2, 0, H)
    return bbx1, bby1, bbx2, bby2

def cutmix_data(x, y, alpha=CUTMIX_ALPHA):
    if alpha <= 0:
        return x, y, None, 1.0
    lam = np.random.beta(alpha, alpha)
    rand_index = torch.randperm(x.size()[0]).to(x.device)
    y_a = y
    y_b = y[rand_index]
    bbx1, bby1, bbx2, bby2 = rand_bbox(x.size(), lam)
    x[:, :, bbx1:bbx2, bby1:bby2] = x[rand_index, :, bbx1:bbx2, bby1:bby2]
    # adjust lambda
    lam_adjusted = 1 - ((bbx2 - bbx1) * (bby2 - bby1) / (x.size()[-1] * x.size()[-2]))
    return x, y_a, y_b, lam_adjusted

def mixup_data(x, y, alpha=MIXUP_ALPHA):
    if alpha <= 0:
        return x, y, None, 1.0
    lam = np.random.beta(alpha, alpha)
    rand_index = torch.randperm(x.size()[0]).to(x.device)
    mixed_x = lam * x + (1 - lam) * x[rand_index, :]
    y_a = y
    y_b = y[rand_index]
    return mixed_x, y_a, y_b, lam

def mixup_criterion(criterion, pred, y_a, y_b, lam):
    return lam * criterion(pred, y_a) + (1 - lam) * criterion(pred, y_b)

# ============================================================
# EMA helper
# ============================================================
def init_ema(model):
    ema = copy.deepcopy(model)
    for p in ema.parameters():
        p.requires_grad_(False)
    return ema

def update_ema(ema_model, model, decay=EMA_DECAY):
    with torch.no_grad():
        msd = model.state_dict()
        for k, v in ema_model.state_dict().items():
            v.copy_(v * decay + (1.0 - decay) * msd[k])

# ============================================================
# Model builder with safe head replacement
# ============================================================
from torchvision.models import ConvNeXt_Base_Weights, EfficientNet_V2_M_Weights, Swin_B_Weights

def build_model(name, num_classes):
    if name == "convnext_base":
        model = models.convnext_base(weights=ConvNeXt_Base_Weights.IMAGENET1K_V1)
        in_f = model.classifier[2].in_features
        model.classifier[2] = nn.Linear(in_f, num_classes)

    elif name == "efficientnet_v2_m":
        model = models.efficientnet_v2_m(weights=EfficientNet_V2_M_Weights.IMAGENET1K_V1)
        in_f = model.classifier[1].in_features
        model.classifier[1] = nn.Linear(in_f, num_classes)

    elif name == "swin_b":
        model = models.swin_b(weights=Swin_B_Weights.IMAGENET1K_V1)
        in_f = model.head.in_features
        model.head = nn.Linear(in_f, num_classes)

    return model.to(DEVICE)

# ============================================================
# Training function for a single model & fold
# ============================================================
def train_single_model(model_name, fold, train_idx, val_idx, df_all):


    print(f"\n--- Training {model_name} fold {fold} ---")
    # Data splits
    train_df = df_all.iloc[train_idx].reset_index(drop=True)
    val_df   = df_all.iloc[val_idx].reset_index(drop=True)

    # Weighted sampler on train for class imbalance
    train_counts = train_df['label'].value_counts().to_dict()
    weights = [1.0/train_counts[lbl] for lbl in train_df['label']]
    sampler = WeightedRandomSampler(weights, num_samples=len(weights), replacement=True)

    train_ds = LungXRayDataset(train_df, transform=train_transforms)
    val_ds   = LungXRayDataset(val_df, transform=val_transforms)
    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, sampler=sampler, num_workers=NUM_WORKERS, pin_memory=(DEVICE=="cuda"))
    val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=(DEVICE=="cuda"))

    model = build_model(model_name, NUM_CLASSES)
    ema_model = init_ema(model)

    # Loss with label smoothing
    criterion = nn.CrossEntropyLoss(label_smoothing=0.1)

    # separate params for head/backbone learning rate
    backbone_params = []
    head_params = []
    for n, p in model.named_parameters():
        if "head" in n or "classifier" in n or "heads" in n:
            head_params.append(p)
        else:
            backbone_params.append(p)

    base_lr_backbone = 3e-5
    base_lr_head = 3e-4
    optimizer = optim.AdamW([
        {'params': backbone_params, 'lr': base_lr_backbone},
        {'params': head_params, 'lr': base_lr_head}
    ], weight_decay=1e-2)

    # SAM wrapper if available
    if SAM_AVAILABLE:
        # torch_optimizer.SAM takes (params, base_optimizer, **kwargs) - adapt usage if API differs
        try:
            base_opt = torch.optim.AdamW  # base for SAM uses same param groups
            optimizer = torch_optim_extra.SAM(model.parameters(), base_optimizer=base_opt, lr=base_lr_head, momentum=0.9, weight_decay=1e-2)
        except Exception:
            # fallback keep optimizer
            pass

    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='max', factor=0.3, patience=3, verbose=True)
    scaler = torch.amp.GradScaler(device="cuda", enabled=(DEVICE=="cuda"))

    best_val_acc = 0.0
    es_counter = 0

    
    for epoch in range(1, EPOCHS+1):
        model.train()
        running_loss = 0.0
        correct = 0
        total = 0
        pbar = tqdm(enumerate(train_loader), total=len(train_loader))
        for i, (imgs, labels) in pbar:
            imgs = imgs.to(DEVICE)
            labels = labels.to(DEVICE)
            optimizer.zero_grad()

            # Choose augment: MixUp or CutMix or none
            r = random.random()
            use_mix = (r < P_MIXUP)
            use_cut = (not use_mix) and (r < P_MIXUP + P_CUTMIX)

            if use_cut:
                imgs_m, y_a, y_b, lam = cutmix_data(imgs.clone(), labels, alpha=CUTMIX_ALPHA)
                with torch.cuda.amp.autocast(enabled=(DEVICE=="cuda")):
                    outputs = model(imgs_m)
                    loss = mixup_criterion(criterion, outputs, y_a, y_b, lam)
            elif use_mix:
                imgs_m, y_a, y_b, lam = mixup_data(imgs, labels, alpha=MIXUP_ALPHA)
                with torch.cuda.amp.autocast(enabled=(DEVICE=="cuda")):
                    outputs = model(imgs_m)
                    loss = mixup_criterion(criterion, outputs, y_a, y_b, lam)
            else:
                with torch.cuda.amp.autocast(enabled=(DEVICE=="cuda")):
                    outputs = model(imgs)
                    loss = criterion(outputs, labels)

            # SAM two-step if available
            if SAM_AVAILABLE and isinstance(optimizer, type(torch_optim_extra.SAM(model.parameters(), base_optimizer=torch.optim.AdamW))):
                # typical SAM usage: first step with loss.backward(), optimizer.first_step()
                # Then second forward/backward and optimizer.second_step()
                # But API may vary; safe fallback: perform single-step update if API unclear.
                try:
                    scaler.scale(loss).backward()
                    optimizer.first_step(zero_grad=True)
                    # second forward-backward
                    with torch.cuda.amp.autocast(enabled=(DEVICE=="cuda")):
                        outputs2 = model(imgs if not use_cut and not use_mix else imgs_m)
                        if use_cut or use_mix:
                            loss2 = mixup_criterion(criterion, outputs2, y_a, y_b, lam)
                        else:
                            loss2 = criterion(outputs2, labels)
                    scaler.scale(loss2).backward()
                    scaler.step(optimizer)
                    scaler.update()
                except Exception:
                    # fallback to normal step
                    scaler.scale(loss).backward()
                    scaler.step(optimizer)
                    scaler.update()
            else:
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()

            # EMA update
            update_ema(ema_model, model, decay=EMA_DECAY)

            running_loss += loss.item()
            # compute simple acc (mixup can distort)
            _, preds = torch.max(outputs, 1)
            correct += (preds == labels).sum().item()
            total += labels.size(0)
            pbar.set_postfix({"loss": running_loss/(i+1), "acc": f"{100*correct/total:.2f}"})

        train_acc = 100 * correct / total
        avg_train_loss = running_loss / len(train_loader)
        print(f"Epoch {epoch} Train loss {avg_train_loss:.4f} Train acc {train_acc:.2f}%")

        # Validation using EMA model for stability
        ema_model.eval()
        val_correct = 0
        val_total = 0
        val_losses = 0.0
        with torch.no_grad():
            for imgs, labels in val_loader:
                imgs = imgs.to(DEVICE)
                labels = labels.to(DEVICE)
                outs = ema_model(imgs)
                loss_v = criterion(outs, labels)
                val_losses += loss_v.item()
                _, preds = torch.max(outs, 1)
                val_correct += (preds == labels).sum().item()
                val_total += labels.size(0)
        val_acc = 100 * val_correct / val_total
        avg_val_loss = val_losses / len(val_loader)
        print(f"Validation - loss: {avg_val_loss:.4f} acc: {val_acc:.3f}%")

        # scheduler step & early stopping
        scheduler.step(val_acc)
        if val_acc > best_val_acc + 1e-5:
            best_val_acc = val_acc
            es_counter = 0
            # Save best EMA weights
            torch.save(ema_model.state_dict(), os.path.join(SAVE_DIR, f"{model_name}_fold{fold}_best.pth"))
            print(f"Saved best {model_name} fold{fold} : {best_val_acc:.4f}%")
        else:
            es_counter += 1
            if es_counter >= EARLY_STOPPING_PATIENCE:
                print("Early stopping triggered.")
                break

    print(f"Finished training {model_name} fold{fold}. Best Val Acc: {best_val_acc:.4f}%")
    return os.path.join(SAVE_DIR, f"{model_name}_fold{fold}_best.pth"), best_val_acc

# ============================================================
# K-Fold training for each model family
# ============================================================
model_families = ["convnext_base", "efficientnet_v2_m", "swin_b"]

from sklearn.model_selection import train_test_split

# One fold (single split)
df_indexed = df_clean.reset_index(drop=True)
labels_arr = df_indexed['label'].map({l: i for i, l in enumerate(CLASSES)}).values

train_idx, val_idx = train_test_split(
    np.arange(len(df_indexed)),
    test_size=0.2,          # 20% for validation
    stratify=labels_arr,
    random_state=SEED
)

all_checkpoints = {m: [] for m in model_families}
fold = 1  # single fold

print(f"\n=== SINGLE FOLD TRAINING (80/20 Split) ===")
for mname in model_families:
    ckpt_path, best_acc = train_single_model(mname, fold, train_idx, val_idx, df_indexed)
    all_checkpoints[mname].append(ckpt_path)


# ============================================================
# Ensemble inference: feature-fusion across model families and folds + TTA
# ============================================================
print("\n--- Ensemble evaluation across folds + TTA ---")

# Build models and load their checkpoint weights (EMA saved)
def load_model_for_inference(model_name, ckpt_path):
    model = build_model(model_name, NUM_CLASSES)
    try:
        state = torch.load(ckpt_path, map_location=DEVICE)
        model.load_state_dict(state)
        print("Loaded", ckpt_path)
    except Exception as e:
        print("Could not load", ckpt_path, e)
    model.eval()
    return model

# Create validation dataset (we will evaluate on full cleaned dataset's holdout by averaging folds)
val_dataset_full = LungXRayDataset(df_indexed, transform=val_transforms)
val_loader_full = DataLoader(val_dataset_full, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS)

# Load all models (families x folds)
ensemble_models = []
for mname in model_families:
    for ckpt in all_checkpoints[mname]:
        ensemble_models.append(load_model_for_inference(mname, ckpt))

print("Total ensemble models loaded:", len(ensemble_models))

# TTA prediction helper: returns (B, num_classes) probs averaged over TTA transforms and models
def predict_with_tta(models, imgs_batch):
    device = imgs_batch.device
    batch_size = imgs_batch.size(0)
    agg_probs = torch.zeros((batch_size, NUM_CLASSES), device=device)
    # for each tta transform, create batch and predict
    for tta in tta_transforms:
        # convert tensor back to uint8 image per sample, apply tta, convert to tensor
        tta_inputs = []
        for i in range(batch_size):
            im = imgs_batch[i].cpu().numpy().transpose(1,2,0)  # C,H,W -> H,W,C (normalized)
            mean = np.array([0.485,0.456,0.406])
            std  = np.array([0.229,0.224,0.225])
            im_unnorm = (im * std + mean)
            im_255 = np.clip(im_unnorm * 255.0, 0, 255).astype(np.uint8)
            aug = tta(image=im_255)
            tta_inputs.append(aug["image"])
        tta_batch = torch.stack(tta_inputs).to(device)
        # sum probabilities from models
        probs_sum = torch.zeros((batch_size, NUM_CLASSES), device=device)
        with torch.no_grad():
            for m in models:
                m.to(device)
                out = torch.softmax(m(tta_batch), dim=1)
                probs_sum += out
        agg_probs += probs_sum / len(models)
    # average across TTA rounds
    agg_probs = agg_probs / len(tta_transforms)
    return agg_probs

# Evaluate across full cleaned dataset
all_labels = []
all_preds = []
all_probs = []
with torch.no_grad():
    for imgs, labels in tqdm(val_loader_full, desc="Ensemble eval"):
        imgs = imgs.to(DEVICE)
        labels = labels.numpy()
        probs = predict_with_tta(ensemble_models, imgs)  # (B,NUM_CLASSES)
        preds = torch.argmax(probs, dim=1).cpu().numpy()
        all_labels.extend(labels.tolist())
        all_preds.extend(preds.tolist())
        all_probs.extend(probs.cpu().numpy())

print("\nFinal Ensemble Classification Report:")
print(classification_report(all_labels, all_preds, target_names=CLASSES))
cm = confusion_matrix(all_labels, all_preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=CLASSES)
disp.plot(cmap="Blues", values_format="d")
import matplotlib.pyplot as plt
plt.title("Final Ensemble Confusion Matrix")
plt.show()

print("Done. Models saved under:", SAVE_DIR)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

def load_training_data_from_csv(file_path='/kaggle/input/training-metrics/training_metrics.csv'):
    """Load training data from CSV file in Kaggle"""
    try:
        df = pd.read_csv(file_path)
        print(f"📊 Models found: {df['model'].unique().tolist()}")
        return df
    except FileNotFoundError:
        print(f"❌ File not found: {file_path}")
        print("Please check the file path and make sure the dataset is added to your Kaggle notebook")
        return None

def create_curves_from_csv(df):
    """Create loss and accuracy curves from CSV data"""
    
    models = df['model'].unique()
    
    for model_name in models:
        # Filter data for current model
        model_data = df[df['model'] == model_name]
        
        epochs = model_data['epoch'].values
        train_loss = model_data['train_loss'].values
        val_loss = model_data['val_loss'].values
        train_acc = model_data['train_acc'].values
        val_acc = model_data['val_acc'].values
        
        # Find best validation accuracy
        best_val_acc = max(val_acc)
        best_epoch = epochs[val_acc.argmax()]
        
        # Create figure with two subplots
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))
        
        # Plot 1: Loss curves - Training: Blue, Validation: Orange
        ax1.plot(epochs, train_loss, 'b-', linewidth=2, label='Train Loss')
        ax1.plot(epochs, val_loss, 'orange', linewidth=2, label='Val Loss')
        ax1.set_xlabel('Epochs')
        ax1.set_ylabel('Loss')
        ax1.set_title(f'{model_name} Fold 1 - Loss Curve')
        ax1.legend()
        ax1.grid(True, alpha=0.3)
        
        # Set y-axis limits for loss
        max_loss = max(max(train_loss), max(val_loss))
        ax1.set_ylim(0, max_loss * 1.1)
        
        # Plot 2: Accuracy curves - Training: Blue, Validation: Orange
        ax2.plot(epochs, train_acc, 'b-', linewidth=2, label='Train Acc')
        ax2.plot(epochs, val_acc, 'orange', linewidth=2, label='Val Acc')
        ax2.set_xlabel('Epochs')
        ax2.set_ylabel('Accuracy (%)')
        ax2.set_title(f'{model_name} Fold 1 - Accuracy Curve\nBest Val Acc: {best_val_acc:.2f}% at Epoch {best_epoch}')
        ax2.legend()
        ax2.grid(True, alpha=0.3)
        
        # Set y-axis limits for accuracy
        ax2.set_ylim(0, 100)
        
        plt.tight_layout()
        plt.show()
        
        print(f"{model_name} - Best Validation Accuracy: {best_val_acc:.2f}% at Epoch {best_epoch}")
        print()

def create_combined_curves_from_csv(df):
    """Create combined curves for all models from CSV data"""
    
    models = df['model'].unique()
    
    # Combined loss comparison
    plt.figure(figsize=(8, 4))
    
    # Loss comparison
    plt.subplot(2, 1, 1)
    colors = ['blue', 'red', 'green', 'purple', 'brown', 'pink']
    
    for i, model_name in enumerate(models):
        model_data = df[df['model'] == model_name]
        plt.plot(model_data['epoch'], model_data['val_loss'], 
                color=colors[i % len(colors)], linewidth=2, 
                label=f'{model_name} Val Loss')
    
    plt.xlabel('Epochs')
    plt.ylabel('Validation Loss')
    plt.title('Model Comparison - Validation Loss')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Accuracy comparison
    plt.subplot(2, 1, 2)
    for i, model_name in enumerate(models):
        model_data = df[df['model'] == model_name]
        plt.plot(model_data['epoch'], model_data['val_acc'], 
                color=colors[i % len(colors)], linewidth=2, 
                label=f'{model_name} Val Acc')
    
    plt.xlabel('Epochs')
    plt.ylabel('Validation Accuracy (%)')
    plt.title('Model Comparison - Validation Accuracy')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

# Main execution - Use this part to load from CSV and plot
if __name__ == "__main__":
    # Load data from CSV
    df = load_training_data_from_csv('/kaggle/input/training-metrics/training_metrics.csv')
    
    # Create individual curves
    create_curves_from_csv(df)
    
    # Create comparison curves
    create_combined_curves_from_csv(df)
    
```
